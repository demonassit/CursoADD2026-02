### Objetivo

| Elemento | Detalle |

| Input | `input_slider` de rango para `YearsAtCompany` (min=0, max=40) |

### Pistas

- El slider de rango funciona igual que el de edad que ya usaste — solo cambia la columna
- Para calcular el porcentaje de rotación: `(d['Attrition'] == 'Yes').mean() * 100`

## Celda 1 — Exploración inicial del dataset

````python
# Instalación y exploración inicial
import pandas as pd
from shiny import App, ui, render


# Cargar el dataset
df = pd.read_csv("../hr_ibm.csv")

# Explorar la variable que usaremos como filtro
print("Exploracion de Años en la Compañia")
print(f"Rango: {df['YearsAtCompany'].min()} - {df['YearsAtCompany'].max()} años")
print(f"Promedio: {df['YearsAtCompany'].mean():.1f} años")
print(f"Mediana: {df['YearsAtCompany'].median():.0f} años")
print()
print("Distribución de los Tramos:")
tramos = [0,2,5,10,20,40]
for i in range(len(tramos)-1):
    sub = df[(df['YearsAtCompany'] >= tramos[i]) & (df['YearsAtCompany'] < tramos[i+1])]
    rot = (sub['Attrition']  == 'Yes').mean()*100
    print(f"{tramos[i]:>2} - {tramos[i+1]:>2} años: {len(sub):>4} empleados|rotación {rot:.1f}% ")
print()
print(f"Tasa de rotación global: { (df['Attrition'] == 'Yes').mean():.1f}%")
````

## Celda 3 — Tu app completa

````python
# ═══════════════════════════════════════════════════════════════
# Escribe aquí tu app completa
# No hay huecos — construye desde cero usando lo aprendido en sesión
# ═══════════════════════════════════════════════════════════════
import pandas as pd
from shiny import App, ui, render


# Cargar el dataset
df = pd.read_csv("../hr_ibm.csv")

app_ui = ui.page_fluid(
    ui.h2("Consulta por Antigüedad"),
    ui.layout_sidebar(
        ui.sidebar(
            ui.h4("Filtro de Antigüedad"),
            ui.input_slider(
                "antig",
                "Años de Antigüedad",
                min=0, max=40, value=[0,40]
            ),
        ),
        ui.output_text("resumen"),
        ui.output_table("tabla"),
    )
)

def server(input, output, session):

    def _filtrar():
        emin, emax = input.antig()
        return df[(df["YearsAtCompany"] >= emin) & (df["YearsAtCompany"] <= emax)]

    @render.text
    def resumen():
        d = _filtrar()
        n = len(d)
        if n == 0:
            return "Sin empleados en este rango"
        pct_rot = (d["Attrition"] == "Yes").mean()*100
        return(f"{n} empleados con {input.antig()[0]} - {input.antig()[1]} " f". Rotación : {pct_rot:.1f}%")

    @render.table
    def tabla():
            cols = ["EmployeeNumber", "Department", "JobRole",
                     "Attrition", "YearsAtCompany", "MonthlyIncome"]
            return _filtrar()[cols].head(10)

app = App(app_ui, server)
# shiny run app.py — o ejecutar app.run() si el entorno lo soporta


import threading

def _iniciar():
    app.run(host="127.0.0.1", port=8000)

hilo = threading.Thread(target=_iniciar, daemon=True)
hilo.start()
print("App corriendo en http://127.0.0.1:8000")
````
